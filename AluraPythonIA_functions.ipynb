```python
import random

def sorteia_sala(alunos, quantidade_de_alunos):
  sala_sorteada = []
  for rodada_do_sorteio in range(int(quantidade_de_alunos) + 1):
    aluno_sorteado = random.choice(alunos)
    sala_sorteada.append(aluno_sorteado)
    alunos.remove(aluno_sorteado)
  return sala_sorteada



```

```python
def remove_itens(lista, itens_a_remover):
  for item in itens_a_remover:
    lista.remove(str(item))
  return lista

```

```python
alunos_sem_sala = [
    "Ana Silva", "João Pedro", "Maria Souza", "Carlos Eduardo", "Beatriz Ribeiro",
    "Lucas Oliveira", "Julia Santos", "Gabriel Costa", "Mariana Almeida", "Felipe Lima",
    "Amanda Pereira", "Rodrigo Gomes", "Larissa Carvalho", "Bruno Martins", "Camila Rodrigues",
    "Diego Cardoso", "Letícia Melo", "Arthur Barbosa", "Isabela Teixeira", "Thiago Ramos"
]

print(f"Sorteio de salas 2027 - Sortearemos {len(alunos_sem_sala)} alunos \n")
print("Alunos que serão sorteados: " + str(alunos_sem_sala))
print("")

sala_1 = sorteia_sala(alunos_sem_sala.copy(), (len(alunos_sem_sala) / 3))
print(f"Sala 1: {sala_1}")
print("")

#remove alunos da sala_1 da lista alunos_sem_sala
alunos_sem_sala = remove_itens(alunos_sem_sala, sala_1)
print("Alunos que serão sorteados na próxima rodada: " + str(alunos_sem_sala))

sala_2 = sorteia_sala(alunos_sem_sala.copy(), (len(alunos_sem_sala)/ 2))
print(f"Sala 2: {sala_2}")
print("")

#remove alunos da sala_2 da lista alunos_sem_sala
alunos_sem_sala = remove_itens(alunos_sem_sala, sala_2)

sala_3 = alunos_sem_sala.copy()
print(f"Sala 3: {sala_3}")
alunos_sem_sala = remove_itens(alunos_sem_sala, sala_3)
```
