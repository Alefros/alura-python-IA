# Trabalhando com Range

```python
#Range de 0 a 4

lista_de_numeros = range(5)
print(lista_de_numeros)

for numero in lista_de_numeros:
  print(numero)

```

```python
#Range entre 2 números

lista_de_numeros = range(5, 10)
print(lista_de_numeros)

for numero in lista_de_numeros:
  print(numero)
```

```python
#Range somente de números pares

lista_de_numeros = range(0, 11, 2)
print(lista_de_numeros)

for numero in lista_de_numeros:
  print(numero)
```

```python
#Range somente números pares, sem usar salto

lista_de_numeros = range(0, 11)

for numero in lista_de_numeros:
  if numero % 2 == 0:
    print(numero)
```
